W przypadku obrazów w odcieniach szarości pojedynczy piksel z zakresu [0; 255] reprezentowany jest jako 8-bitowa liczba bez znaku.
Pewnym rozszerzeniem analizy sposobu reprezentacji obrazu może być następujący eksperyment.
Załóżmy, że z każdego z 8 bitów możemy stworzyć pojedynczy obraz binarny (ang. _bit-plane slicing_).
Dla obrazka _100zloty.jpg_ (https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/100zloty.jpg), przekonwertowanego na skalę odcieni szarości, stwórz 8 obrazów, z których każdy powinien zawierać jedną płaszczyznę bitową.
Podpowiedź $-$ warto sprawdzić, jak realizuje się bitowe operacje logiczne.
Zastosowanie takiej dekompozycji obrazu pozwala na analizę ,,ważności'' poszczególnych bitów.
Jest to użyteczne w kwantyzacji, ale także w kompresji.
Uwaga, proszę odpowiednio przeskalować wyświetlane wartości (do zakresu 0-255 lub ustawić zakres na 0-1), w przeciwnym przypadku cały obraz będzie czarny.

W drugim etapie zadania proszę spróbować odtworzyć obraz oryginalny z mniejszej liczby obrazów binarnych.
Warto zacząć od dwóch najbardziej znaczących bitów, a później dodawać kolejne.
Należy utworzyć co najmniej trzy wersje zrekonstruowanych obrazów.
Podpowiedź $-$ rekonstrukcja obrazu to mnożenie przez odpowiednią potęgę liczby 2 (przesunięcie bitowe) oraz dodawanie.

```python
import cv2
import os
import requests
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/vision-agh/poc_sw/master/02_Point/'
fileName = '100zloty.jpg'
if not os.path.exists(fileName) :
    r = requests.get(url + fileName, allow_redirects=True)
    open(fileName, 'wb').write(r.content)
```

```python
img = cv2.imread(fileName, cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.title('obraz oryginalny')
plt.show()
```

```python
planes = []
fig, axes = plt.subplots(2, 4, figsize=(16, 8))

for bit in range(8):
    plane = (img >> bit) & 1
    planes.append(plane)

    ax = axes.flat[bit]
    ax.imshow(plane, cmap='gray', vmin=0, vmax=1)
    ax.set_title(f'bit {bit} (waga {2 ** bit})')

plt.show()
# wyzsze bity zachowuja glowne ksztalty, nizsze pokazuja drobne roznice jasnosci
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, count in zip(axes, [2, 4, 6]):
    reconstructed = img * 0
    for bit in range(7, 7 - count, -1):
        reconstructed += planes[bit] * (2 ** bit)

    ax.imshow(reconstructed, cmap='gray', vmin=0, vmax=255)
    ax.set_title(f'{count} najważniejszych bitów')

plt.show()
# wiecej bitow daje wiecej poziomow szarosci i dokladniejsza rekonstrukcje
```
